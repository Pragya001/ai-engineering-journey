```python
symbols = "@#$^!*"

codes = []
for s in symbols:
    codes.append(ord(s))
codes
```

```python
#listcomps
symbols = "@#$^!*"

codes = [ord(s) for s in symbols]
codes
```

```python
#listcomps instead of maps
symbols = "@#$^!*"

codes = [ord(s) for s in symbols if ord(s) > 40]
codes
```

```python
#lambda
symbols = "@#$^!*"

codes = list(filter(lambda c: c > 40, map(ord, symbols)))
codes
```

```python
#cartesian product through listcomps
ranks = ['A', 'K', 'Q']
suits = ['S', 'H', 'D', 'C']

product = [(r, s) for r in ranks for s in suits]
product
```

```python
#genexp for tuple generation

symbols = "@#$^!*"
tuple(ord(s) for s in symbols)
```

```python
#genexp for array generation
import array 

symbols = "@#$^!*"
array.array('I', (ord(s) for s in symbols))
```

```python
#cartesian Product using genexp

sizes = ['S', 'M', 'L']
colors = ['Black', 'White']

for item in (f'{size} {color}' for size in sizes for color in colors):
    print(item)
```

```python
#tuple as immutable
a = [23, 'abc', [1, 2]]
b = [23, 'abc', [1, 2]]

print(a == b)
b[-1].append(3)
print(a==b)
```

```python
#Check Hashable tuple or not

def fixed(o):
    try:
        hash(o)
    except TypeError:
        return False
    return True
tf = (2, 'abs', (12,4))
tm = (2, 'abs', [12,4])

print(fixed(tf))
print(fixed(tm))
```

```python
# Using * for unpacking

x = divmod(4, 8)
print(x)

a = (4, 8)
x = divmod(*a)
print(x)

quotient, remainder = divmod(*a)
print(quotient)
print(remainder)


```

```python
# Using * for excess items

a, b, *rest, d = range(6)
print(a, b, rest, d)

a, b, *rest, d = range(4)
print(a, b, rest, d)

a, b, *rest = range(2)
print(a, b, rest)
```

```python
#Unpacking with * in Function Calls and Sequence Literals

def fun(a, b, c, d, *rest):
    return a, b, c, d, rest

print(fun(*[1,2], 3, 4, *range(5, 8)))
```

```python
# Use of * in defining list

list = [*range(3, 9), 9, 10]
print(list)
```
